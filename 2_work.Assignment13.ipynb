```python

import pandas as pd

# Step 1: Load cleaned dataset
df = pd.read_csv(r"C:\Users\dell\Downloads\teaching_prices_cleaned_easy.csv")

print(df.head())
print(df.info())

# Step 2: Check missing values
print(df.isnull().sum())

# Step 3: Calculate mean and median
stats = df.groupby("item")["price"].agg(
    ["mean", "median"]
)

print(stats.round(2))

# Step 4: First-to-last percentage change
df["date"] = pd.to_datetime(df["date"])

weekly = df.groupby(
    ["item", "date"]
)["price"].mean().reset_index()

first = weekly.groupby("item").first()["price"]
last = weekly.groupby("item").last()["price"]

change = ((last - first) / first) * 100

stats["percentage_change"] = change

print(stats.round(2))

# Step 5: Test flagged price 9999
sweet = df[df["item"] == "Sweetener D"].copy()

sweet["original_number"] = pd.to_numeric(
    sweet["price_original"].astype(str).str.replace(",", ""),
    errors="coerce"
)

without_flag = sweet["price"]

with_flag = sweet["price"].fillna(
    sweet["original_number"]
)

print("Without flagged value:")
print("Mean:", without_flag.mean())
print("Median:", without_flag.median())

print("With flagged value:")
print("Mean:", with_flag.mean())
print("Median:", with_flag.median())

# Step 6: Save statistics
stats.to_csv("assignment3_statistics.csv")

print("Assignment 3 statistics completed!")
```
